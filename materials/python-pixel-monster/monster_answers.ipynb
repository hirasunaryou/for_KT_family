## MISSION 1: 1ピクセルから、モンスター誕生

```python
from PIL import Image
from IPython.display import display

design = (
    '................',
    '....##....##....',
    '...#HH#..#HH#...',
    '..#BBBB##BBBB#..',
    '.#BBBBBBBBBBBB#.',
    '.#BBBWWBBWWBBB#.',
    '.#BBBWEBBEWBBB#.',
    '.#BBCBBBBBBCBB#.',
    '.#BBBBB##BBBBB#.',
    '..#BBBBBBBBBB#..',
    '...#BBBBBBBB#...',
    '..#BB######BB#..',
    '.#BBB#....#BBB#.',
    '.####......####.',
    '................',
    '................',
)
palette = {
    '.': (0, 0, 0, 0),
    '#': (32, 44, 71, 255),
    'B': (113, 87, 223, 255),
    'H': (186, 169, 255, 255),
    'W': (250, 250, 255, 255),
    'E': (20, 25, 40, 255),
    'C': (255, 120, 165, 255),
}

def draw_monster(design, palette):
    height = len(design)
    width = len(design[0])
    image = Image.new("RGBA", (width, height), (0, 0, 0, 0))
    for y, row in enumerate(design):
        for x, symbol in enumerate(row):
            image.putpixel((x, y), palette[symbol])
    return image

monster = draw_monster(design, palette)
print(monster.size, monster.mode)
print("左上:", monster.getpixel((0, 0)))
display(monster.resize((320, 320), Image.Resampling.NEAREST))
```

```python
from monster_tools import show_pixels

# 背景の市松模様は、透明な場所を見やすくする表示用
show_pixels(monster)
```

## MISSION 2: 色違いと、おばけを作る

```python
def replace_color(image, old, new):
    result = image.copy()
    for y in range(result.height):
        for x in range(result.width):
            if result.getpixel((x, y)) == old:
                result.putpixel((x, y), new)
    return result

red = replace_color(monster, palette["B"], (235, 80, 75, 255))
blue = replace_color(monster, palette["B"], (40, 165, 220, 255))
ghost = replace_color(monster, palette["B"], (113, 87, 223, 90))
```

## MISSION 4: 自分のモンスターチーム

```python
variants = [monster, red, blue]
team = Image.new("RGBA", (56, 20), (238, 243, 250, 255))
for i, image in enumerate(variants):
    x = 2 + i * 18
    team.alpha_composite(image, dest=(x, 2))

# 半透明の体を、違う背景に重ねる
night = Image.new("RGBA", (20, 20), (24, 35, 58, 255))
night.alpha_composite(ghost, dest=(2, 2))
```

```python
from monster_tools import show_pixels

show_pixels(team, scale=8)
show_pixels(night)
```

## MISSION 5: モンスター自動生成工場

```python
import random

def make_monster(seed=0):
    rng = random.Random(seed)
    body = (rng.randint(50, 220), rng.randint(50, 220), rng.randint(50, 220), 255)
    outline = (32, 44, 71, 255)
    image = Image.new("RGBA", (16, 16), (0, 0, 0, 0))
    for y in range(3, 12):
        left = rng.randint(2, 5)
        for x in range(left, 8):
            image.putpixel((x, y), body)
            image.putpixel((15 - x, y), body)
        image.putpixel((left - 1, y), outline)
        image.putpixel((16 - left, y), outline)
    for x in [5, 6, 9, 10]:
        for y in [5, 6]:
            image.putpixel((x, y), (250, 250, 255, 255))
    for x in [6, 9]:
        image.putpixel((x, 6), (20, 25, 40, 255))
    for x in [7, 8]:
        image.putpixel((x, 9), outline)
    for x in [5, 6, 9, 10]:
        image.putpixel((x, 12), outline)
    return image

collection = [make_monster(seed) for seed in range(9)]
```

```python
from monster_tools import show_gallery

show_gallery(collection, columns=3)
```
